```python

import math

# ---------------- Part A: Machine Initialization ----------------
machines = ['A', 'B', 'C']
M = len(machines)
N_play = 10
c = 2                      # exploration constant (UCB1 uses sqrt(2 ln t / N))

# Set AUTO = True to use the suggested reward pattern instead of typing rewards
AUTO = False
pattern = {'A': [3, 4, 2, 5], 'B': [6, 7, 5, 8], 'C': [4, 3, 5, 4]}
pattern_idx = {m: 0 for m in machines}

# ---------------- Part B: Knowledge tables ----------------
N   = {m: 0   for m in machines}   # times each machine played
R   = {m: 0   for m in machines}   # total reward from each machine
Q   = {m: 0.0 for m in machines}   # estimated value
UCB = {m: 0.0 for m in machines}   # UCB score

def get_reward(machine):
    """Part D: reward collection"""
    if AUTO:
        r = pattern[machine][pattern_idx[machine] % len(pattern[machine])]
        pattern_idx[machine] += 1
        print(f"Enter reward: {r}   (auto)")
        return r
    return float(input("Enter reward: "))

total_reward = 0

for t in range(1, N_play + 1):
    # ---------------- Part C: UCB calculation & selection ----------------
    if t <= M:
        # initial test: play each machine once to initialise N, R, Q
        chosen = machines[t - 1]
        print(f"\nPlay {t}: initial test -> machine {chosen}")
    else:
        for m in machines:
            UCB[m] = Q[m] + math.sqrt(c * math.log(t) / N[m])
        chosen = max(machines, key=lambda m: UCB[m])
        print(f"\nPlay {t}: UCB = { {m: round(UCB[m], 3) for m in machines} }")
        print(f"Play with machine {chosen}")

    # ---------------- Part D: reward ----------------
    reward = get_reward(chosen)

    # ---------------- Part E: update N, R, Q ----------------
    N[chosen] += 1
    R[chosen] += reward
    Q[chosen] = Q[chosen] + (reward - Q[chosen]) / N[chosen]   # incremental sample average
    total_reward += reward
    print(f"N[{chosen}] = {N[chosen]},  Q[{chosen}] = {Q[chosen]:.3f}")

# UCB values after the final play (using t = N_play)
for m in machines:
    UCB[m] = Q[m] + math.sqrt(c * math.log(N_play) / N[m])

# ---------------- Part F: Display results ----------------
print("\n" + "=" * 55)
print(f"{'Machine':<10}{'N':<8}{'R (total)':<12}{'Q':<10}{'UCB':<10}")
for m in machines:
    print(f"{m:<10}{N[m]:<8}{R[m]:<12}{Q[m]:<10.3f}{UCB[m]:<10.3f}")
print("=" * 55)
print("Total reward received:", total_reward)
best = max(machines, key=lambda m: Q[m])
print(f"Machine with highest Q-value: {best}")
```
